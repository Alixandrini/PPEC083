```python
import folium
from folium.plugins import SideBySideLayers

# Coordenadas centrais de Salvador - BA
lat, lon = -12.9777, -38.5016
zoom_inicial = 13

# 1. Cria o mapa base
m = folium.Map(
    location=[lat, lon],
    zoom_start=zoom_inicial,
    tiles="OpenStreetMap",
    control_scale=True,
)

# 2. Camada WMS da Esquerda (Salvador 1998)
url_wms_1998 = (
    "https://mapas.conder.ba.gov.br/arcgis/services/MOSAICO/Salvador_1998/ImageServer/WMSServer"
)
layer_left = folium.raster_layers.WmsTileLayer(
    url=url_wms_1998,
    layers="0",
    name="Mosaico Salvador 1998",
    fmt="image/png",
    transparent=False,
    version="1.3.0",
    attr="CONDER (1998)",
)

# 3. Camada WMS da Direita (RMS / Recôncavo 2010)
url_wms_2010 = (
    "https://mapas.conder.ba.gov.br/arcgis/services/MOSAICO/RMS_Reconcavo_2010/ImageServer/WMSServer"
)
layer_right = folium.raster_layers.WmsTileLayer(
    url=url_wms_2010,
    layers="0",
    name="Mosaico RMS Recôncavo 2010",
    fmt="image/png",
    transparent=False,
    version="1.3.0",
    attr="CONDER (2010)",
)

# 4. Adiciona as camadas ao mapa
layer_left.add_to(m)
layer_right.add_to(m)

# 5. Adiciona o controle Side-by-Side
sbs = SideBySideLayers(layer_left=layer_left, layer_right=layer_right)
sbs.add_to(m)

# 6. Salva o HTML
m.save("split_map_salvador_1998_2010.html")
print("Arquivo salvo: split_map_salvador_1998_2010.html")
```
