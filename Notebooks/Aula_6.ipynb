 # Reconhecimento de Imagens

 O reconhecimento de imagens, frequentemente chamado de classificação de imagens ou classificação de cenas na literatura de visão computacional, é a tarefa de atribuir um único rótulo categórico a uma imagem inteira. Dado um tile de imagem de satélite ou aérea, um classificador pode rotulá-lo como “floresta”, “residencial”, “industrial” ou “agrícola”. Diferentemente da segmentação, que produz um rótulo para cada pixel, a classifi cação de imagens trata o tile inteiro como uma unidade e pergunta: “Qual é o uso ou cobertura do solo dominante ou mais representativo nesta cena?”


Essa  classificação  no  nível  de  tile  é  uma  das  tarefas  mais  fundamentais  em  IA  geoespacial  e  serve como base para tarefas mais complexas abordadas em capítulos posteriores.

Abaixo estão os principais exemplos de aplicação prática dessa abordagem em sensoriamento remoto e gestão territorial:

1. **Monitoramento e Triagem Rápida de Desmatamento e Degradação**
* **Como funciona**: Grandes faixas de imagens orbitais (como CBERS-4A ou Sentinel-2) são particionadas em uma grade de recortes regulares (ex.: $64 \times 64$ ou $128 \times 128$ pixels).
* **Mecanismo Convolucional**: O modelo não olha apenas se o pixel é verde ou marrom; ele reconhece a transição textural de bordas retas (clareiras de corte raso) ou cicatrizes lineares (estradas vicinais para extração de madeira) em meio à cobertura contínua de floresta.
* **Aplicação Prática**: Geração de alertas rápidos de mudança de uso do solo para direcionar a fiscalização de órgãos ambientais sem a necessidade de processar segmentações vetorizadas completas em tempo real.

2. **Identificação e Mapeamento de Sistemas Agrícolas Específicos**
* **Como funciona**: Diferenciação entre padrões agrícolas com respostas espectrais semelhantes em nível de pixel isolado, mas estruturas espaciais distintas.
* **Mecanismo Convolucional**:Pivôs Centrais de Irrigação: A convolução extrai as feições circulares e o contraste das bordas em relação ao relevo circundante.Culturas Perenes vs. Pastagem: Talhões de citros ou cafezais exibem espaçamentos regulares em linha e texturas rugosas que uma camada convolucional captura facilmente, distinguindo-os de uma pastagem homogênea ou lavoura temporária.
* **Aplicação Prática**: Inventário automatizado de infraestrutura de irrigação para outorga de recursos hídricos e estimativas de safra por cooperativas agropecuárias.

3. **Delimitação e Tipificação de Zonas Urbanas**
* **Como funciona**: Caracterização do tecido urbano a partir do padrão de ocupação do solo.
* **Mecanismo Convolucional**: A rede aprende filtros que detectam alta densidade de bordas (telhados, arruamento ortogonal) vs. grandes estruturas contínuas (galpões logísticos) ou áreas com arruamento irregular e alta densidade de edificações sem recuo (assentamentos precários/favelas).
* **Aplicação Prática**:Atualização de planos diretores municipais.Mapeamento de vazios urbanos e identificação de frentes de expansão urbana informal para planejamento de infraestrutura e saneamento.

4. **Triagem e Avaliação de Danos em Desastres Naturais**
* **Como funciona**: Classificação do status de células territoriais antes e depois de eventos extremos (inundações, deslizamentos de terra ou rompimento de barragens).
* **Mecanismo Convolucional**: Comparação de recortes para rotulá-los entre classes como Área Intacta, Área Inundada ou Estrutura Danificada. A rede identifica a ruptura de padrões conhecidos (vias interrompidas por sedimentos, deposição lamacenta sobre malha urbana)
* **Aplicação Prática**: Suporte à Defesa Civil nas primeiras horas após o desastre, priorizando setores geográficos críticos para resgate e envio de suprimentos.

5. **Mineração e Extração Mineral (Legal e Ilegal)**
* **Como funciona**: Varredura sistemática de bacias hidrográficas e áreas florestais para detecção de cavas de garimpo ou pilhas de rejeito.
* **Mecanismo Convolucional**: A convolução detecta a quebra súbita da textura florestal associada a padrões de solo revolvido, presença de tanques de sedimentação com coloração turva e maquinário.
* **Aplicação Prática**: Apoio a auditorias de regularidade ambiental e fiscalização de garimpos clandestinos em terras indígenas e unidades de conservação.

### 1. Aprendizado por Transferência (*Transfer Learning*)



* **Tradução Conceitual:** Transferência de Aprendizado / Aprendizado por Transferência.
* **O Conceito:** Redes Neurais Convolucionais profundas exigem centenas de milhares (ou milhões) de imagens para aprender representações visuais robustas a partir do zero (*from scratch*). No *Transfer Learning*, utiliza-se uma rede previamente treinada em uma base massiva de imagens (frequentemente o ImageNet, com mais de 1 milhão de fotos) e reaproveitam-se esses pesos para uma nova tarefa — no caso, a classificação de imagens de satélite.


* **Como Funciona na Prática:**
* **Extrator de Feições Congelado (*Feature Extractor*):** As camadas iniciais de uma CNN aprendem feições universais de baixo nível (arestas, gradientes de cor, texturas e linhas ortogonais). Essas camadas podem ser mantidas congeladas.
* **Ajuste Fino (*Fine-Tuning*):** Substitui-se apenas a última camada densa/linear (a cabeça de classificação, *head*) para coincidir com o número de classes de interesse (por exemplo, as 10 classes do EuroSAT). Os pesos da rede são então ajustados suavemente com uma taxa de aprendizado menor.




* **Vantagem em Sensoriamento Remoto:** Reduz drasticamente o tempo de treinamento e viabiliza modelos de alta acurácia mesmo com amostragens reduzidas de imagens orbitais.


### 2. ResNet (*Residual Networks*)



* **Tradução Conceitual:** Redes Residuais.
* **O Problema Resolvido:** Historicamente, ao empilhar dezenas de camadas convolucionais, as redes sofriam com a degradação de desempenho e o desaparecimento do gradiente (*vanishing gradient*), onde os sinais de correção calculados pelo *backpropagation* se tornavam próximos de zero antes de alcançar as camadas iniciais.
* **A Solução das Conexões Residuais (*Skip Connections / Shortcut*):**
* Em vez de forçar as camadas a aprenderem um mapeamento direto $H(x)$, a ResNet formula o bloco para aprender um resíduo:

$$F(x) = H(x) - x \implies H(x) = F(x) + x$$


* O termo $+ x$ representa uma ligação direta que pula uma ou mais camadas convolucionais. Se uma camada não acrescentar ganho preditivo, ela pode simplesmente convergir seus pesos para zero, mantendo a identidade intacta ($H(x) = x$).


* **No Notebook (`resnet50`):** Utiliza uma variante de 50 camadas profundas dividida em blocos residuais do tipo *bottleneck*. É uma arquitetura de referência robusta e estável para benchmarks de cobertura do solo.


### 3. EfficientNet



* **Tradução Conceitual:** Redes Convolucionais Eficientes por Escalonamento Composto.
* **O Problema Resolvido:** Tradicionalmente, para melhorar o desempenho de uma rede, ampliava-se apenas uma dimensão: ou a profundidade (número de camadas, como na ResNet), ou a largura (quantidade de canais/filtros), ou a resolução espacial de entrada. O aumento desordenado de qualquer uma dessas variáveis de forma isolada causa saturação rápida de ganho e eleva os custos computacionais.
* **A Abordagem do Escalonamento Composto (*Compound Scaling*):**
* A EfficientNet balanceia uniformemente profundidade ($d$), largura ($w$) e resolução de entrada ($r$) sob uma restrição de recursos computacionais fixos, utilizando um coeficiente composto $\phi$:

$$\text{Profundidade: } d = \alpha^\phi, \quad \text{Largura: } w = \beta^\phi, \quad \text{Resolução: } r = \gamma^\phi$$


* A arquitetura base (B0) emprega blocos convolucionais invertidos com conexões residuais (*MBConv* — *Mobile Inverted Bottleneck Convolution*), combinados com mecanismos de atenção leves (*Squeeze-and-Excitation*).


* **No Notebook (`efficientnet_b0`):** Representa a versão de entrada da família EfficientNet. Entrega níveis de acurácia comparáveis ou superiores aos da ResNet-50 com significativamente menos parâmetros treináveis e menor pegada de memória.

---

### Comparativo Direto entre as Duas Abordagens no Notebook

| Característica | ResNet-50 | EfficientNet-B0 |
| --- | --- | --- |
| **Mecanismo Central** | Conexões de salto residuais (*Skip Connections*) | Escalonamento composto balanceado + Blocos *MBConv*<br> |
| **Volume de Parâmetros** | Elevado (~25 milhões de parâmetros) | Reduzido (~5.3 milhões de parâmetros) |
| **Custo Computacional / Memória** | Maior consumo de GPU por época | Muito leve, ideal para execuções ágeis e inferência rápida |
| **Comportamento em Imagens Menores (64×64)** | Mantém boa capacidade de generalização espacial | Altamente eficiente, mas pode exigir atenção na resolução mínima aceita pelas reduções de escala sucessivas |

````python
import os
from geoai.utils import download_file
from geoai.recognize import (
    load_image_dataset,
    train_image_classifier,
    predict_images,
    evaluate_classifier,
    plot_training_history,
    plot_confusion_matrix,
    plot_predictions,
    push_classifier_to_hub,
    predict_images_from_hub,
)
````

# Download the EuroSAT RGB Dataset

````python
url = "https://data.source.coop/opengeos/geoai/EuroSAT-RGB.zip"
data_dir = download_file(url)
````

````python
print(f"Dataset directory: {data_dir}")
print(f"Files: {sorted(os.listdir(data_dir))}")
````

# Explorar o Conjunto de Dados

Desempacota os metadados do dicionário: listas de nomes de classes, caminhos, rótulos numéricos e o mapa de classe para índice.   plt.subplots(2, 5, ...): Cria uma grade com 2 linhas e 5 colunas para acomodar exatamente as 10 classes do EuroSAT.   O loop varre cada classe, recupera a primeira ocorrência visual (labels.index(...)), abre a imagem, desativa os eixos cartesianos (ax.axis("off")) e atribui o nome da classe ao título.   Renderiza a figura consolidada

````python
dataset_info = load_image_dataset(data_dir)
print(f"Classes ({len(dataset_info['class_names'])}): {dataset_info['class_names']}")
print(f"Total images: {len(dataset_info['image_paths'])}")
````

````python
import matplotlib.pyplot as plt
from PIL import Image

class_names = dataset_info["class_names"]
image_paths = dataset_info["image_paths"]
labels = dataset_info["labels"]
class_to_idx = dataset_info["class_to_idx"]

fig, axes = plt.subplots(2, 5, figsize=(20, 8))

for idx, class_name in enumerate(class_names):
    ax = axes[idx // 5, idx % 5]
    # Find first image of this class
    img_idx = labels.index(class_to_idx[class_name])
    img = Image.open(image_paths[img_idx])
    ax.imshow(img)
    ax.set_title(class_name, fontsize=12)
    ax.axis("off")

plt.suptitle("Sample Image from Each Class", fontsize=14)
plt.tight_layout()
plt.show()
````

# Treinar um Classificador ResNet-50

Executa o fine-tuning da arquitetura ResNet-50:   
* ```data_dir```: Diretório base das imagens.   
* ```model_name="resnet50"```: Seleciona o backbone convolucional residual.   
* ```num_epochs=5```: Número de ciclos completos de treinamento sobre os dados.   
* ```batch_size=32```: Tamanho do lote de imagens processado a cada iteração do gradiente.  
*```learning_rate=1e-3```: Taxa de aprendizado inicial do otimizador.  
* ```image_size=64```: Dimensão de redimensionamento dos patches ($64 \times 64$ pixels).   
* ```in_channels=3```: Especifica 3 canais de entrada (RGB).   
* ```pretrained=True```: Inicializa o modelo com pesos pré-treinados do ImageNet (transfer learning).   
* ```output_dir```: Pasta onde checkpoints e métricas serão persistidos.   
* ```num_workers=4 e seed=42```: Define threads paralelas para carregar dados e fixa a semente de números aleatórios para reprodutibilidade.  

````python
result = train_image_classifier(
    data_dir=data_dir,
    model_name="resnet50",
    num_epochs=5,
    batch_size=32,
    learning_rate=1e-3,
    image_size=64,
    in_channels=3,
    pretrained=True,
    output_dir="image_recognition_output/resnet50",
    num_workers=4,
    seed=42,
)
````

# Plotar o Histórico de Treinamento

````python
fig = plot_training_history("image_recognition_output/resnet50/models")
plt.show()
````

# Avaliar no Conjunto de Teste

Lê os registros de log da pasta de saída e plota as curvas de perda (loss) e acurácia ao longo das 5 épocas.

````python
eval_result = evaluate_classifier(
    model=result["model"],
    dataset=result["test_dataset"],
    class_names=result["class_names"],
)
````

# Plotar a Matriz de Confusão

Plota a matriz de confusão normalizada por linha (valores entre 0 e 1), permitindo enxergar a taxa de acerto e erro percentual de cada classe independentemente do balanceamento

````python
fig = plot_confusion_matrix(
    eval_result["confusion_matrix"],
    result["class_names"],
)
plt.show()
````

````python
fig = plot_confusion_matrix(
    eval_result["confusion_matrix"],
    result["class_names"],
    normalize=True,
)
plt.show()
````

# Visualizar as Predições

Sorteia 10 índices pseudoaleatórios (rng.sample) dentro da partição de teste.   


```predict_images(...)```: Executa o forward pass nas imagens selecionadas retornando as classes preditas e o vetor de probabilidades (softmax).  

 
```plot_predictions(...)```: Gera a visualização gráfica comparando lado a lado a imagem, a classe real (ground truth), a classe predita e a barra de confiança do modelo.   

````python
import random

test_dataset = result["test_dataset"]
test_paths = test_dataset.image_paths
test_labels = test_dataset.labels
n_samples = min(10, len(test_paths))

rng = random.Random(42)
sample_indices = rng.sample(range(len(test_paths)), k=n_samples)
sample_paths = [test_paths[i] for i in sample_indices]
sample_labels = [test_labels[i] for i in sample_indices]

pred_result = predict_images(
    model=result["model"],
    image_paths=sample_paths,
    class_names=result["class_names"],
    image_size=64,
    in_channels=3,
)

fig = plot_predictions(
    image_paths=sample_paths,
    predictions=pred_result["predictions"],
    true_labels=sample_labels,
    class_names=result["class_names"],
    probabilities=pred_result["probabilities"],
)
plt.show()
````
